```python
# Correlation of all numeric columns with Price_USD_M
corr = df.corr(numeric_only=True)["Price_USD_M"].drop("Price_USD_M")

print("Pearson Correlation with Price_USD_M:\n")

for column, value in corr.items():
    print(f"{column:<20}: {value:+.3f}")

print('Strongest Positive Feature : Area_SqFt (',round(max(corr),3),')')
print('Strongest Negative Feature : Distance_City_KM (',round(min(corr),3),')')
```

```python
df["Value_group"] = np.where(df["Price_USD_M"] > 1.5,"High Value","Moderate Value")
print(df['Value_group'].head())
```

```python
count = df["Value_group"].value_counts()
print(count)

mean1 = df.groupby(['Value_group'])[['Area_SqFt','Distance_City_KM','School_Rating']].mean()
print(mean1)
```
