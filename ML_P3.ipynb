```python
df1 = pd.merge(df_shipments, df_fleet, on='Vehicle_ID')
df2 = pd.merge(df1, df_weather, on='Region_ID')
print(df2)
```

```python
print("No of Rows: ",df2.shape[0])
print("No of Columns: ",df2.shape[1])
print("\nColumn Names: ")

for i in df2.columns:
    print(i)
```

```python
X = df2.drop(columns = ['Shipment_ID','Vehicle_ID','Region_ID'], axis=1)
y = df2['Is_Delayed']
print(X)
print(y)
```

```python
avg = df2.groupby(df2['Is_Delayed'])[['Distance_KM','Maintenance_Score']].mean()
print(avg)
print(f"- Mean Distance            : Delayed = {avg.loc['Yes', 'Distance_KM']:.2f} | On Time = {avg.loc['No', 'Distance_KM']:.2f}")

print(f"- Mean Maintenance Score   : Delayed = {avg.loc['Yes', 'Maintenance_Score']:.2f} | On Time = {avg.loc['No', 'Maintenance_Score']:.2f}")
```

```python
df2['Prediction'] = df2.apply(lambda row: 'Yes' if ((row['Distance_KM'] > 400 and row['Maintenance_Score'] < 60)
                                                    or (row['Weather_Condition'] in ['Storm', 'Heavy Rain']))
                                                else 'No', axis=1)
print(df2[['Is_Delayed','Prediction']])
```

```python
correct_pred = (df2['Is_Delayed'] == df2['Prediction']).sum()
total_pred = len(df2)
print(f"Correct Predictions: {correct_pred}")
print(f"Total Predictions: {total_pred}")
accuracy = correct_pred/total_pred*100
print(f"Accuracy: {accuracy:.2f}%")
```

```python
false_positive = ((df2["Prediction"] == "Yes") & (df2["Is_Delayed"] == "No")).sum()
false_negative = ((df2["Prediction"] == "No") & (df2["Is_Delayed"] == "Yes")).sum()
print(f"False Positives : {false_positive}")
print(f"False Negatives : {false_negative}")
```
