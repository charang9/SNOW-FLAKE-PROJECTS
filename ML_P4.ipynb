```python
median = df3.groupby(df3['Is_Fraud'])[['Tx_Amount','Terminal_Risk_Score']].median()
print("Group Medians:")
print('- Median Tx Amount   : Fraud = ',median.loc[1,'Tx_Amount'],'| Legitamate = ',median.loc[0,'Tx_Amount'],sep = '')
print('- Terminal Risk Score: Fraud = ',median.loc[1,'Terminal_Risk_Score'],'   | Legitamate = ',median.loc[0,'Terminal_Risk_Score'],sep = '')
```

```python

df3["Prediction"] = df3.apply(
    lambda row: 1
    if (row["Tx_Amount"] > 1000 and row["Tx_Hour"] in [1, 2, 3, 4])
    or (row["High_Risk_IP_Flag"] == 1)
    else 0,
    axis=1
)
print(df3[['Tx_ID','Is_Fraud','Prediction']])
```

```python
correct_pred = (df3['Is_Fraud'] == df3['Prediction']).sum()
total_pred = len(df3)
accuracy = (correct_pred / total_pred)*100
print(f'Accuracy: {accuracy:.2f}%')
```
