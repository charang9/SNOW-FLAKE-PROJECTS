```python
import pandas as pd
data = {
    "Customer_ID": ["C101", "C102", "C103", "C104", "C105", "C106", "C107"],
    "Account_Age_Days": [120, 450, 90, 300, 210, 60, 500],
    "Monthly_Spend": [45.5, 120.0, 25.0, 85.0, 60.0, 15.0, 200.0],
    "Support_Calls": [4, 0, 5, 1, 2, 6, 1],
    "Inactivity_Days": [35, 5, 40, 12, 20, 45, 8],
    "Churn": ["Yes", "No", "Yes", "No", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(' Number of Rows    : ',df.shape[0],'\n','Number of Columns : ',df.shape[1])
print("\nColumns:")
for i in df.columns:
  print(i)
```

```python
X = df.drop(columns = ['Customer_ID','Churn'])
y = df['Churn']

print('Features','\n',X,'\n\nLabels\n',y)
```

```python
count = df['Churn'].value_counts()
print(count)
```

```python
print('Average Inactivity Days:')

avg = df.groupby(df['Churn'])['Inactivity_Days'].mean()

print('Churned Customers   : ',avg['Yes'],'\n','Retained Customers : ',avg['No'],sep='')
```

```python
df['Prediction'] = df.apply(lambda row:'Yes' if row['Inactivity_Days'] > 30 and row['Monthly_Spend'] < 50
                                           else 'No',axis = 1)

print(df[['Customer_ID','Churn','Prediction']])
```

```python
correct_pred = (df['Churn'] == df['Prediction']).sum()
total_pred = len(df)
Accuracy = (correct_pred/total_pred)*100
print("Correct Predictions : ",correct_pred)
print("Total Predictions   : ",total_pred)
print(f"Accuracy : {Accuracy:.0f}%")
```

```python
print("=" * 10, "CUSTOMER CHURN ANALYSIS", "=" * 10)

print(f"Total Customers        : {len(df)}")
print(f"Churned Customers      : {count['Yes']}")
print(f"Retained Customers     : {count['No']}")

print("\nAverage Inactivity")
print(f"Churned                : {avg['Yes']} days")
print(f"Retained               : {avg['No']} days")

print(f"\nRule-Based Accuracy    : {Accuracy:.0f}%")

print("\nNew Customer A Pred    : No")
print("New Customer B Pred    : No")
```
