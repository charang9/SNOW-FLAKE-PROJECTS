```python
import pandas as pd
import numpy as np

patient_data = {
    "Patient_ID": ["P_501","P_502","P_503","P_504","P_505","P_506","P_507","P_508","P_509","P_510"],
    "Heart_Rate_BPM": [72,125,95,68,140,88,75,130,92,70],
    "Systolic_BP": [120,165,138,115,180,130,122,170,135,118],
    "Oxygen_Sat_Pct": [98,89,94,99,85,95,97,88,93,98],
    "Age": [35,68,45,28,75,52,40,62,50,30],
    "Triage_Category": ["General","ICU","Urgent","General","ICU","Urgent","General","ICU","Urgent","General"]
}

df = pd.DataFrame(patient_data)

print(df)
```

```python
print("Supervised Paradigm : Multi-Class Classification")
print("Target Type         : Discrete Categorical")
```

```python
X = df.drop(columns=["Patient_ID","Triage_Category"])
y = df["Triage_Category"]

print("Features (X):")
print(X.head())

print("\nTarget OR Labels (y):")
print(y.head())
```

```python
class_count = df["Triage_Category"].value_counts()
class_percent = df["Triage_Category"].value_counts(normalize=True)*100

print(class_count)
print('\n',class_percent,sep = '')
print("\nClass Distribution",sep = '')

for category in class_count.index:
    print(f"{category}: {class_count[category]} ({class_percent[category]:.1f}%)")
```

```python
group_avg = df.groupby("Triage_Category")[["Heart_Rate_BPM","Systolic_BP","Oxygen_Sat_Pct"]].mean()

print(group_avg.round(2))
```
