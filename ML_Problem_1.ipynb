# Smart Employee Attendance Prediction

**Week 4 – Day 16**

This notebook analyzes employee attendance data using Pandas and compares a rule-based approach with Machine Learning.


```python
import pandas as pd

data = {
    "Employee_ID": ["E101", "E102", "E103", "E104", "E105"],
    "Distance_KM": [5, 18, 7, 22, 10],
    "Travel_Time_Min": [20, 65, 30, 75, 40],
    "Previous_Late_Count": [1, 5, 0, 7, 2],
    "Weather": ["Clear", "Rain", "Clear", "Rain", "Cloudy"],
    "Transport": ["Bus", "Bus", "Car", "Bus", "Bike"],
    "Late": ["No", "Yes", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

print(df)

print(f'Number of rows: {df.shape[0]}{"\n"}Number of columns: {df.shape[1]}')
```

```python
late_count = df["Late"].value_counts()

print(late_count)

```

Task 5

```python
avg = df.groupby('Late')['Travel_Time_Min'].mean()
```

```python
print(avg)
```

```python
print(" Late Employees:",avg['Yes'],'\n','Not Late Employees:',avg['No'])
```

```python
df["Prediction"] = df.apply(
    lambda row: "Yes" if row["Distance_KM"] > 15 and row["Travel_Time_Min"] > 60
                else "No",
    axis=1
)

print(df[["Employee_ID", "Late", "Prediction"]])
```

```python
correct_predictions = (df["Late"] == df["Prediction"]).sum()
total_predictions = len(df)
accuracy = (correct_predictions / total_predictions) * 100

print(f"Correct Predictions : {correct_predictions}")
print(f"Total Predictions   : {total_predictions}")
print(f"Rule-Based Accuracy : {accuracy:.0f}%")
```

```python
distance = 20
travel_time = 70

if distance > 15 and travel_time > 60:
    prediction = "Yes"
else:
    prediction = "No"

print("New Employee 1")
print(f"Distance        : {distance} KM")
print(f"Travel Time     : {travel_time} Minutes")
print(f"Prediction      : {prediction}")
```
